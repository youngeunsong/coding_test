# 출력
영문 소문자 'q'가 입력될 때까지 입력한 문자를 계속 출력한다.  
x  
b  
k  
d  
l  
q  

```python
character = 'a'
n = 0 

while n == 0 :
    character = input()
    print(character)

    if character == 'q':
        n += 1
```
