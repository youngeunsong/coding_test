# 문제 설명 

본 문제는 python 의 빠른 기초 학습을 위해 설계된 문제로서 python 코드 제출을 기준으로 설명되어 있습니다. 
------
이번에는 특수문자 출력에 도전하자!!

다음 문장을 출력하시오.


"!@#$%^&*()'

(단, 큰따옴표와 작은따옴표도 함께 출력한다.)

 

출력 형식에 필요한 따옴표와 출력할 문자인 따옴표를 구분하기 위하여

\" 또는 \' 를 이용하여 출력할 수 있다.


** 주의 : 본 화면에서 복사하여 붙여넣기하면 제대로 되지 않을 수 있으니 직접 작성해야한다.

```python
print('"'+"!@#$%^&*()"+"'")
```

```python
print('\"!@#$%^&*()\'')
```
