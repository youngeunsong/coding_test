# 2751: 수 정렬하기 2

## 문제
N개의 수가 주어졌을 때, 이를 오름차순으로 정렬하는 프로그램을 작성하시오.

## 입력
첫째 줄에 수의 개수 N(1 ≤ N ≤ 1,000,000)이 주어진다. 둘째 줄부터 N개의 줄에는 수가 주어진다. 이 수는 절댓값이 1,000,000보다 작거나 같은 정수이다. 수는 중복되지 않는다.

## 출력
첫째 줄부터 N개의 줄에 오름차순으로 정렬한 결과를 한 줄에 하나씩 출력한다.

## 예제 입력 1
```
5
5
4
3
2
1
```

## 예제 출력 1 
```
1
2
3
4
5
```

````python
# 내가 작성한 코드: 예제는 통과했으나 제출 시 틀렸다고 나옴. 그러나 틀린 테스트 케이스가 무엇인지 알려주지 않아서 수정이 여의치 않음. 
import sys

### 백준 제출 시 제거
import io
input_data = """10
5
2
3
1
4
2
3
5
1
7"""

sys.stdin = io.StringIO(input_data)
###
input = sys.stdin.readline

# 입력 받기 
n = int(input())

# 카운트 리스트 정의 
counts = [0] * 1000001

# 입력 받으면서 정렬
for _ in range(n):
    num = int(input())
    counts[num] += 1

# 오름차순으로 정렬된 수 출력
for i in range(len(counts)):
    for j in range(counts[i]):
        print(i)
````
