# 알고리즘 수업 - 삽입 정렬 3
## 문제
오늘도 서준이는 삽입 정렬 수업 조교를 하고 있다. 아빠가 수업한 내용을 학생들이 잘 이해했는지 문제를 통해서 확인해보자.

N개의 서로 다른 양의 정수가 저장된 배열 A가 있다. 삽입 정렬로 배열 A를 오름차순 정렬할 경우 정렬 과정에서 배열 A가 배열 B와 같은 경우가 발생하는지 확인해 보자. 초기 상태 배열 A도 정렬 과정에서 발생 가능한 경우로 생각하자.

크기가 N인 배열에 대한 삽입 정렬 의사 코드는 다음과 같다.

```
insertion_sort(A[1..N]) { # A[1..N]을 오름차순 정렬한다.
    for i <- 2 to N {
        loc = i - 1;
        newItem = A[i];

        # 이 지점에서 A[1..i-1]은 이미 정렬되어 있는 상태
        while (1 <= loc and newItem < A[loc]) {
            A[loc + 1] <- A[loc];
            loc--;
        }
        if (loc + 1 != i) then A[loc + 1] = newItem;
    }
}
```

## 입력
첫째 줄에 배열 A, B의 크기 N(5 ≤ N ≤ 10,000)이 주어진다.

다음 줄에 서로 다른 배열 A의 원소 A1, A2, ..., AN이 주어진다. (1 ≤ Ai ≤ 109)

다음 줄에 배열 B의 원소 B1, B2, ..., BN이 주어진다. (1 ≤ Bi ≤ 109)

## 출력
삽입 정렬로 배열 A를 오름차순 정렬하는 과정에서 배열 A가 배열 B와 같은 경우가 발생하면 1, 아니면 0을 출력한다.

## 예제 입력 1 
```
5
4 5 1 3 2
1 3 4 5 5
```

## 예제 출력 1
1

````python
# 내가 작성한 코드
import sys

### 백준 제출 시 제거
import io
input_data = '''5
4 5 1 3 2
1 3 4 5 5
'''
sys.stdin = io.StringIO(input_data)
###
input = sys.stdin.readline

# 입력 받기
n = int(input()) # 배열 a, b 크기
a = list(map(int, input().split())) # 배열 a
b = list(map(int, input().split()))

# 삽입 정렬 함수 구현 
# insertion_sort(A[1..N]) { # A[1..N]을 오름차순 정렬한다.
#     for i <- 2 to N {
#         loc = i - 1;
#         newItem = A[i];

#         # 이 지점에서 A[1..i-1]은 이미 정렬되어 있는 상태
#         while (1 <= loc and newItem < A[loc]) {
#             A[loc + 1] <- A[loc];
#             loc--;
#         }
#         if (loc + 1 != i) then A[loc + 1] = newItem;
#     }
# }
def insertion_sort(a, b): 
    for i in range(1, n):
        loc = i - 1
        newItem = a[i]

        while 0 <= loc and newItem < a[loc]:
            a[loc + 1] = a[loc]
            loc -= 1

            # a와 b 행렬 일치하는 순간 생기는 지 확인 
            if a == b: 
                return print(1)
        
        if loc + 1 != i: 
            a[loc + 1] = newItem

            # a와 b 행렬 일치하는 순간 생기는 지 확인 
            if a == b: 
                return print(1)
    return print(0)

insertion_sort(a, b)
````
