# 성적이 낮은 순서대로 학생 출력하기
## 제한
* 시간 제한: 1초
* 메모리 제한: 128 MB

## 문제 설명
* N명의 학생 정보: 학생의 이름, 학생의 성적
* 성적이 낮은 순서대로 학생의 이름을 출력하시오

## 입력 조건
* 첫 번째 줄: 학생의 수 N (1 <= N <= 100,000)
* 두 번째 줄 ~ N + 1 번째 줄: 학생의 이름을 나타내는 문자열 A, 학생의 성적을 나타내는 정수 B가 공백으로 구분. 문자열 A의 길이와 학생의 성적은 100 이하의 자연수

## 출력 조건
모든 학생의 이름을 성적이 낮은 순서대로 출력. 성적이 동일한 학생들의 이름 순서는 자유롭게 출력 가능 

## 입력 예시
```
2 
홍길동 95
이순신 77
```

## 출력 예시
이순신 홍길동 

````python
# 모범 답안: 기본 정렬 함수 이용. 
# 학생의 정보가 최대 100,000개이므로 기본 정렬 함수 가능
# 튜플 자료형을 활용 
import sys
import io
input_data = '''2 
홍길동 95
이순신 77
'''
sys.stdin = io.StringIO(input_data)
input = sys.stdin.readline

# N 입력 받기
n = int(input())

# N명의 학생 정보를 입력받아 리스트에 저장
array = []

for i in range(n):
    a, b = input().split()
    b = int(b) # 성적은 정수형으로 변환 
    array.append((a, b))

# 키를 이용하여 점수를 기준으로 정렬
array = sorted(array, key = lambda student: student[1])

# 정렬이 수행된 결과를 출력
for student in array:
    print(student[0], end = ' ')
````

````python
# 내가 작성한 답안 
import sys
import io
input_data = '''2 
홍길동 95
이순신 77
'''
sys.stdin = io.StringIO(input_data)
input = sys.stdin.readline

n = int(input())
array = []

for _ in range(n):
    a, b = input().split()
    b = int(b)
    array.append((a, b))
# print(array)

def returnB(data):
    return data[1]

array.sort(key = returnB )

for i in range(n):
    print(array[i][0], end = ' ')
# sortedDict = sorted(dictionary, key = )
````
