# 알고리즘 수업 - 선택 정렬 2
## 문제 설명
오늘도 서준이는 선택 정렬 수업 조교를 하고 있다. 아빠가 수업한 내용을 학생들이 잘 이해했는지 문제를 통해서 확인해보자.

N개의 서로 다른 양의 정수가 저장된 배열 A가 있다. 선택 정렬로 배열 A를 오름차순 정렬할 경우 K 번 교환이 발생한 직후의 배열 A를 출력해 보자.

크기가 N인 배열에 대한 선택 정렬 의사 코드는 다음과 같다.

```
selection_sort(A[1..N]) { # A[1..N]을 오름차순 정렬한다.
    for last <- N downto 2 {
        A[1..last]중 가장 큰 수 A[i]를 찾는다
        if (last != i) then A[last] <-> A[i]  # last와 i가 서로 다르면 A[last]와 A[i]를 교환
    }
}
```

## 입력
첫째 줄에 배열 A의 크기 N(5 ≤ N ≤ 10,000), 교환 횟수 K(1 ≤ K ≤ N)가 주어진다.

다음 줄에 서로 다른 배열 A의 원소 A1, A2, ..., AN이 주어진다. (1 ≤ Ai ≤ 109)

## 출력
 K 번 교환이 발생한 직후의 배열 A를 한 줄에 출력한다. 교환 횟수가 K 보다 작으면 -1을 출력한다.

## 예제 입력 1
```
5 2
3 1 2 5 4
```

## 예제 출력 1
2 1 3 4 5

````python
# 내가 작성한 코드
# 뼈대가 되는 알고리즘 
# selection_sort(A[1..N]) { # A[1..N]을 오름차순 정렬한다.
#     for last <- N downto 2 {
#         A[1..last]중 가장 큰 수 A[i]를 찾는다
#         if (last != i) then A[last] <-> A[i]  # last와 i가 서로 다르면 A[last]와 A[i]를 교환
#     }
# }

import sys
### 백준 제출 시 삭제할 파트
import io
input_data = """5 2
3 1 2 5 4
"""
sys.stdin = io.StringIO(input_data)
### 여기까지 제거 
input = sys.stdin.readline

# 배열의 크기 N, 교환 횟수 K번
n, k = map(int, input().split())

# 배열 a
a = list(map(int, input().split()))

# 선택 정렬 구현하기 
def selection_sort(array):
    count = 0 

    for i in range(n - 1, 0, -1):
        last = i 

        # 최댓값 찾기
        for j in range(last + 1):
            if array[j] > array[last]:
                last = j
        
        # 최댓값과 스왑
        if last != i:
            array[i], array[last] = array[last], array[i]
            count += 1
        
        # k번 교환 후 원소 출력    
        if count == k:
            result = ""
            for i in range(n):
                result += str(array[i]) + " "
            return(print(result))
            # for i in range(n):
            #     print(array[i], end = " ")
    # 만약 k번 미만으로 교환하면 -1 출력
    return(print(-1))


# 선택 정렬 실행 
selection_sort(a)
````
