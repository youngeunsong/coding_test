# 바닥 공사
가로의 길이가 N, 세로의 길이가 2인 직사각형 형태의 얇은 바닥. 이 바닥을 1 x 2의 덮개, 2 x 1의 덮개, 2 x 2의 덮개를 이용해 채우고자 한다. 이 때 바닥을 채우는 모든 경우의 수를 구하시오. 

예) 2(세로) x 3(가로): 총 5가지 경우의 수

## 입력 조건
첫째 줄: N (1 <= N <= 1000)

## 출력 조건
경우의 수를 796,796으로 나눈 나머지를 출력하시오

```python
n = int(input())

d = [0] * (n+1)
d[1] = 1
d[2] = 3

# 점화식: a[N] = a[N-1] + a[N - 2] * 2
for i in range(3, n + 1):
    d[i] = (d[i - 1] + d[i - 2] * 2) % 796796

print(d[n])
```
