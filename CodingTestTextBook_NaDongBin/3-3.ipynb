# 숫자 카드 게임
시간 제한 1초, 메모리 제한 128 MB

게임 룰  
가장 낮은 숫자가 쓰인 카드 한 장을 뽑는 게임. 단, 아래 룰을 지켜야 함. 

1. N x M 배치의 카드 (N = 행의 갯수, M = 열의 갯수)
1. 먼저 뽑고자 하는 카드가 포함되어 있는 행을 선택
1. 그다음 선택된 행에 포함된 카드들 중 가장 숫자가 낮은 카드 뽑기 
1. 따라서 처음 카드를 골라낼 행을 선택할 때, 이후에 해당 행에서 가장 숫자가 낮은 카드를 뽑을 것으로 고려하여 최종적으로 가장 높은 숫자의 카드를 뽑을 수 있도록 전략 세우기 

## 입력 조건
* 첫째 줄에 N, M 공백을 두고 주어짐. 1 <= N, M <= 100
* 둘째 줄부터 N개의 줄에 걸쳐 각 카드에 적힌 숫자가 주어진다. 

## 출력 조건
* 룰에 맞게 선택된 카드에 적힌 숫자 출력

## 입력 예시 1
3 3  
3 1 2  
4 1 4  
2 2 2  

## 출력 예시 1
2

## 입력 예시 2
2 4  
7 3 1 8  
3 3 3 4 

## 출력 예시 2
3


```python
# 내가 작성한 코드 
n, m = map(int, input().split())
cards = []
for _ in range(n):
    cards.append(list(map(int, input().split())))
# print(cards)  

# 각 행의 최솟값을 찾고, 이 중 가장 큰 값을 출력하면 될 것. 
minCards = 1

for i in range(n):
    # print(cards[i])
    minRow = min(cards[i])
    minCards = minRow if minRow > minCards else minCards 
print(minCards)
```

```python
del min # 실수로 파이썬의 내장 함수와 동일한 변수를 만들어서 'TypeError: 'int' object is not callable' 오류 발생 시 해당 변수를 제거 후 다시 코드 실행하여 문제 해결 가능 
```

```python
# 모범 답안
# N, M을 공백으로 구분하여 입력 받기
n, m = map(int, input().split())

result = 0

# 한 줄 씩 입력받아 확인
for i in range(n):
    data = list(map(int, input().split()))
    # 현재 줄에서 가장 작은 수 찾기
    min_value = min(data)
    # 가장 작은 수들에서 최댓값 찾기
    result = max(result, min_value)

print(result) # 최종 답안 출력
```
