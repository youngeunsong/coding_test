# 재귀 함수를 이용한 피보나치 수열 구현 함수 
x가 커짐에 따라 수행 시간이 급속히 길어짐
<!-- 빅오 표기법: $$O(2^N)$$ -->
빅오 표기법: O(2<sup>N</sup>)

```python
def fibo(x):
    if x == 1 or x == 2:
        return 1
    return fibo(x -1) + fibo(x - 2)

print(fibo(4))
```

# 다이나믹 프로그래밍을 적용한 피보나치 수열 구현 함수
메모이제이션 활용
탑다운 방식

```python
# 한 번 계산된 결과를 메모이제이션 하기 위한 메모리 초기화
d = [0] * 100

# 피보나치 함수를 재귀함수로 구현(탑다운 다이나믹 프로그래밍)
def fibo(x): 
    # 종료 조건 (1 혹은 2일 때 1 반환)
    if x == 1 or x == 2:
        return 1
    
    # 이미 게산한 적 있는 문제라면 그대로 반환
    if d[x] != 0: 
        return d[x]
    
    # 아직 계산하지 않는 식이라면 점화식에 따라 계산
    d[x] = fibo(x - 1) + fibo(x - 2)
    return d[x]

print(fibo(99))
```

바텀업 방식

```python
# 앞서 계산된 결과를 초기화하기 위한 DP 테이블 초기화
d = [0] * 100

# 첫 번째 피보나치 수와 두 번째 피보나치 수는 1
d[1] = 1
d[2] = 1
n = 99

# 반복문으로 수열 구현
for i in range(3, n + 1):
    d[i] = d[i - 1] + d[i -2]

print(d[n])
```
