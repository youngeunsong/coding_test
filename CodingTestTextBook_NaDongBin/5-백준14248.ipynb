# 점프 점프 (BFS)
## 문제
영우는 개구리다 개굴개굴개굴

영우는 지금 n개의 돌이 일렬로 놓여있는 돌다리에 있다. 그리고 돌다리의 돌에는 숫자가 하나씩 적혀있다. 영우는 이 숫자가 적혀있는 만큼 왼쪽이나 오른쪽으로 점프할 수 있는데, 이때 돌다리 밖으로 나갈 수는 없다.

영우는 이 돌다리에서 자기가 방문 가능한 돌들의 개수를 알고 싶어한다. 방문 가능하다는 것은 현재위치에서 다른 돌을 적절히 밟아 해당하는 위치로 이동이 가능하다는 뜻이다.

현재 위치가 주어졌을 때, 영우가 방문 가능한 돌들의 개수를 출력하시오.

## 입력
첫 번째 줄에는 돌다리의 돌 개수 n이 주어진다.(1≤n≤100,000) 돌의 번호는 왼쪽부터 1번에서 n번이다. 다음 줄에는 그 위치에서 점프할 수 있는 거리 Ai가 주어진다.(1≤Ai≤100,000)

다음 줄에는 출발점 s가 주어진다.(1≤s≤n)

## 출력
영우가 방문 가능한 돌들의 개수를 출력하시오.

```
5
1 4 2 2 1
3
```

````python
# 내가 작성한 코드 초안: 오답이었으나 모범 답안 참고하여 성공시킴  
import sys
### 여기서부턴 백준 제출 시 제거
import io
input_data = """5
1 4 2 2 1
3
"""
sys.stdin = io.StringIO(input_data)
### 여기까지 제거
input = sys.stdin.readline

# 총 돌의 갯수    
n = int(input())

# 돌 별로 이동 가능한 수 저장
graph = list(map(int, input().split()))
visited = [False] * n 
# 출발점
start = int(input()) - 1 # start 값을 행렬 인덱스에 맞게 변경 

# bfs 함수 정의
from collections import deque
def bfs(node): 
    queue = deque()
    queue.append(node)
    visited[node] = True 

    # 큐가 빌 때까지 반복
    while queue:
        node = queue.popleft()

        # for 문을 이용해 좌 혹은 우 혹은 양방향 모두 방문 가능한 지 확인.
        for d in [-graph[node], graph[node]]:
            nextNode = node + d

            if 0 <= nextNode < n and not visited[nextNode]:
                visited[nextNode] = True
                queue.append(nextNode)

bfs(start)
print(visited.count(True))
````
