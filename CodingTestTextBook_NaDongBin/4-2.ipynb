# 시각
정수 N이 입력되면 00시 00분 00초부터 N시 59분 59초까지의 모든 시각 중 3이 하나롣 포함되는 모든 경우의 수를 구하기. 

예시: N = 1 -> 00시 00분 03초, 00시 13분 300초 등이 해당함. 

# 입력 조건
* 첫째 줄에 정수 N 입력. 0 <= N <= 23

# 출력 조건
* 3이 하나라도 포함되어 있는 경우의 수 출력

# 입력 예시 
5

# 출력 예시
11475

```python
# 내가 작성한 코드: 모범 답안과 거의 동일해서 기분이 좋다 
n = int(input())

count = 0 # 3이 들어가는 경우의 수 세기 

for hh in range(n+1):
    for mm in range(60):
        for ss in range(60):
            currTime = str(hh) + str(mm) + str(ss)

            if currTime.find("3") != -1:
                count += 1

print(count)
```
