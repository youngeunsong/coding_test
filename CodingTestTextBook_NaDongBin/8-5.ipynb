# 1로 만들기
정수 x가 주어질 때 가능한 연산 
1. x 가 5의 배수: 5로 나누기
1. x가 3의 배수: 3으로 나누기
1. x가 2의 배수: 2로 나누기
1. x - 1

정수 x가 주어졌을 때 연산 4개를 적절히 사용해서 1로 만들려고 함. 필요한 최소 연산 횟수를 구하시오. 

예) x = 26일 때 3번의 연산이 최솟값
1. 26 -1 = 25
1. 25 / 5 = 5
3. 5 / 5 = 1

## 입력 조건
* 첫째 줄: 정수 x (1 <= x <= 30000)

## 출력 조건
* 첫째 줄에 연산을 하는 횟수의 최솟값을 출력    

```python
n = int(input() )
dp = [0] * 30001
count = 0

for i in range(2, n + 1):
    # 현재의 수에서 1을 빼는 경우
    dp[i] = dp[i - 1] + 1 
    # 현재의 수가 2의 배수인 경우
    if i % 2 == 0:
        dp[i] = min(dp[i], dp[i // 2] + 1) 
    # 현재의 수가 3의 배수인 경우
    if i % 3 == 0:
        dp[i] = min(dp[i], dp[i // 3] + 1)
    # 현재의 수가 5의 배수인 경우
    if i % 5 == 0:
        dp[i] = min(dp[i], dp[i // 5] + 1)

print(dp[n])
```
