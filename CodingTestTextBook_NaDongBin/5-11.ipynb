# 미로 탈출 (153 P)
* N x M 크기의 미로   
* 주인공의 위치는 (1, 1)
* 출구 위치: (N, M)
* 한 번에 한 칸씩 이동 가능 
* 괴물이 있는 부분: 0, 없는 부분: 1
* 미로는 반드시 탈출할 수 있는 형태로 제시됨
* 탈출에 필요한 최소 칸의 갯수 구하시오
* 칸을 셀 때 시작 칸과 마지막 칸을 모두 포함하여 구하시오 

## 입력 조건
* 첫째 줄에 정수 N, M이 주어짐. 공백으로 구분
* 둘째 줄부터 N개의 줄에 M개의 정수가 주어짐. 공백없이 제공. 또한 시작칸과 마지막 칸은 항상 1

## 출력 조건
* 첫째 줄에 최소 이동 칸의 개수를 출력 

## 입력 예시
```
5 6
101010
111111
000001
111111
111111
```

## 출력 예시
10 

````python
# 모범 답안
from collections import deque 

# 입력값 받아오기 
n, m = map(int, input().split())
graph = []
for _ in range(n):
    graph.append(list(map(int, input())))

# 이동할 네 방향 정의
dx = [-1, 1, 0, 0]
dy = [0, 0, -1, 1]

# BFS 소스 코드 구현 
def bfs(x, y):
    # 큐 구현을 위해 deque 라이브러리 활용
    queue = deque()
    queue.append((x, y))

    # 큐가 빌 때까지 반복
    while queue:
        x, y = queue.popleft()

        # 현재 위치에서 네 방향으로의 위치 확인
        for i in range(4):
            nx = x + dx[i]
            ny = y + dy[i]

            # 미로 찾기 공간을 벗어나는 경우는 무시
            if nx >= n or nx < 0 or ny >= m or ny < 0:
                continue
            # 벽(괴물)인 경우 무시
            if graph[nx][ny] == 0: 
                continue

            # 해당 노드를 처음 방문하는 경우에만 최단 거리 기록
            if graph[nx][ny] == 1:
                graph[nx][ny] = graph[x][y] + 1
                queue.append((nx, ny))
    # 가장 오른쪽 아래까지의 최단 거리 반환
    return graph[n -1][m - 1]

# BFS를 수행한 결과 출력
print(bfs(0, 0))
 
````
