# 떡볶이 떡 만들기 
떡을 자르는 높이 H를 지정. 서로 다른 떡을 해당 높이 만큼 자르고 남은 것을 합쳐서 손님이 가져가는 떡의 길이가 결정됨. 떡이 해당 높이보다 짧다면 잘리지 않음. 

손님이 요청한 총 길이가 M일 때 최소 M만큼의 떡을 얻기 위해 절단기에 설정할 수 있는 높이의 최댓값을 구하는 프로그램을 작성하시오. 

## 입력 조건
* 첫째 줄: 떡의 개수 N, 요청한 떡의 길이 M (1 <= N <= 1,000,000, 1 <= M <= 2,000,000,000)
* 둘째 줄: 각 떡의 높이. 총합은 항상 M 이상이므로 언제나 손님은 떡을 구매 가능. 각 높이는 10억 이하이고 0 이상인 정수. 

## 출력 조건
* 적어도 M만큼의 떡을 가져가기 위해 절단기에 설정할 수 있는 높이의 최댓값 출력

## 입력 예시
```
4 6
19 15 10 17
```

## 출력 예시
15

````python
# 입력 받기
import sys
import io

input_data = """4 6
19 15 10 17"""
sys.stdin = io.StringIO(input_data)
input = sys.stdin.readline

n, m = map(int, input().split())
heights = list(map(int, input().split()))

start = 0
end = max(heights)

# 이진 탐색 수행
result = 0 
while start <= end:
    mid = (start + end) // 2

    # 남는 떡 길이 계산
    remained = 0 
    for x in heights:
        if x > mid:
            remained += (x - mid)
    
    # 이진 탐색
    # 떡의 양이 부족하면 더 많이 자르기 (왼쪽 탐색)
    if remained < m:
        end = mid - 1
    else: # 떡의 양이 m 이상이면 result에 기록 
        result = mid 
        start = mid + 1

print(result)
````
