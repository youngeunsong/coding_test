# 실전 문제: 부품 찾기
N개의 전자 부품이 있는 매장. 정수 형태의 고유 번호 부여됨. 손님이 M개 종류의 부품을 대량으로 구매하겠다고 견적서 요청. 이 때 가게 안에 부품이 모두 있는 지 확인하는 프로그램 작성하기. 



## 입력 조건
* 첫째 줄: 정수 N (1 <= N <= 1,000,000)
* 둘째 줄: 공백으로 구분된 N개의 정수(총 보유 부품 번호). N과 동일한 값의 범위 가짐. 
* 셋째 줄: 정수 M (1 <= N <= 1,00,000) 
* 넷째 줄: 공백으로 구분된 M개의 정수. N과 동일한 값의 범위 가짐. 

## 출력 조건
* 첫째 줄에 공백으로 구분하여 각 부품이 존재하면 yes, 없으면 no를 출력. 

## 입력 예시
```
5
8 3 7 9 2
3 
5 7 9
```

## 출력 예시
no yes yes

````python
# 내가 작성한 코드 
import sys
import io
input_data = """5
8 3 7 9 2
3 
5 7 9"""
sys.stdin = io.StringIO(input_data)
input = sys.stdin.readline

# 입력 받기
n = int(input())
storedParts = list(map(int, input().split()))
storedParts.sort() # 이진 탐색 적용을 위해 정렬 
m = int(input())
targetParts = list(map(int, input().split()))

# 이진 탐색 함수 정의
def binary_search(array, target, start, end): 
    if start > end:
        return None
    
    mid = (start + end) // 2

    if target == array[mid]:
        return mid
    elif target < array[mid]:
        return binary_search(array, target, start, mid - 1)
    elif target > array[mid]: 
        return binary_search(array, target, mid + 1, end)

for i in range(m):
    if binary_search(storedParts, targetParts[i], 0,  n - 1) != None:
        print("yes", end = " ")
    else: 
        print("no", end = " ")
````

````python
# 교재에 나온 모범 답안
# 이진 탐색
def binary_search(array, target, start, end):
    while start <= end: 
        mid = (start + end) // 2

        # 찾은 경우 중간점 인덱스 반환 
        if array[mid] == target:
            return mid
        # 중간점의 값 > 타겟
        elif array[mid] > target:
            end = mid - 1
        # 중간점의 값 < 타겟
        elif array[mid] < target:
            start = mid + 1
    return None


import sys
import io
input_data = """5
8 3 7 9 2
3 
5 7 9"""
sys.stdin = io.StringIO(input_data)
input = sys.stdin.readline

# 입력 받기
n = int(input())
storedParts = list(map(int, input().split()))
storedParts.sort() # 이진 탐색 적용을 위해 정렬 
m = int(input())
targetParts = list(map(int, input().split()))

# 손님이 확인 요청한 번호 하나씩 확인
for i in targetParts:
    # 해당 부품이 존재하는 지 확인
    result = binary_search(storedParts, i, 0, n - 1)

    if result != None:
        print('yes', end = ' ')
    else: 
        print('no', end = ' ')

````

````python
# 계수 정렬 활용
import sys
import io
input_data = """5
8 3 7 9 2
3 
5 7 9"""
sys.stdin = io.StringIO(input_data)
input = sys.stdin.readline

# 입력 받기
n = int(input())
array = [0] * 1000001

for i in input().split():
    array[int(i)] = 1

m = int(input())
targetParts = list(map(int, input().split()))

# 손님 요청 부품 존재 여부 확인
for i in range(m):
    # 해당 부품 존재 여부 확인
    if array[targetParts[i]] == 1:
        print("yes", end = " ")
    else:
        print("no", end = " ")

````

````python
# 집합 자료형 이용: 단순히 각 부품 존재 여부만 알면 되기에 이용 가능한 방법
# 계수 정렬 활용
import sys
import io
input_data = """5
8 3 7 9 2
3 
5 7 9"""
sys.stdin = io.StringIO(input_data)
input = sys.stdin.readline

n = int(input())
array = set(map(int, input().split())) # 집합 자료형으로 중복 제거
m = int(input())
requests = list(map(int, input().split()))

for i in requests:
    if i in array:
        print("yes", end = " ")
    else:
        print("no", end = " ")
````
