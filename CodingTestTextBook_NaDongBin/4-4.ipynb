# 게임 개발
* 시간 제한: 1초
* 메모리 제한: 128 MB

맵: N x M 크기  
각 칸은 육지 혹은 바다  
캐릭터 방향: 동서남북 중 하나
위치 표시: (A, B) (A: 북쪽으로부터 떨어진 칸의 갯수, B: 서쪽으로부터 떨어진 칸의 갯수)  
캐릭터: 상하좌우 이동 가능, 바다는 갈 수 없음.  

매뉴얼
1. 현재 위치에서 현재 방향을 기준으로 왼쪽 방향(반시계 방향으로 90도 회전한 방향)으로 회전  
1. 캐릭터의 바로 왼쪽 방향에 아직 가보지 않은 칸이 존재한다면 왼쪽 방향으로 회전 후 왼쪽으로 한 칸을 전진. 모두 가 본 칸이란 왼쪽 방향으로 회전만 수행하고 1단계로 돌아가기
1. 만약 네 방향 모두 이미 가본 칸이거나 바다 칸이라면, 바라보는 방향을 유지한 채 한 칸 뒤로 가고 1단계로 돌아가기. 단, 이때 뒤쪽 방향이 바다인 칸이라 뒤로 갈 수 없는 경우에는 움직임을 멈춘다. 

이 매뉴얼에 따라 이동 후 최종적으로 캐릭터가 방문한 칸의 수 출력하기  

## 입력 조건
* 첫째 줄: 맵의 세로 크기 N, 가로 크기 M을 공백으로 구분하여 출력 (3 <= N, M <= 50)
* 둘째 줄: 게임 캐릭터 위치 좌표 (A, B)와 바라보는 방향이 서로 공백으로 구분하여 주어짐. d는 다음과 같이 4가지 존재. 

1. 0: 북쪽
1. 1: 동쪽
1. 2: 남쪽
1. 3: 서쪽 

* 셋째 줄부터 맵의 정보 제공. N개의 줄에 맵의 상태가 북쪽부터 남쪽 순서로, 각 줄에서는 서쪽부터 동쪽 순서대로 주어짐. 맵의 외곽은 항상 바다
1. 0: 육지
1. 1: 바다

* 처음에 게임 캐릭터가 위치한 칸의 상태는 항상 육지

## 출력 조건
* 첫째 줄에 이동을 마친 후 캐릭터가 방문한 칸의 수를 출력한다. 

## 입력 예시
4 4  # 4*4 맵 생성    
1 1 0  # (1, 1)에 북쪽(0)을 바라보고 서 있는 캐릭터  
1 1 1 1 # 첫 줄은 모두 바다    
1 0 0 1   
1 1 0 1  
1 1 1 1   

## 출력 예시
3

```python
# 내가 작성한 코드 -> 현재 출력 예시대로 답이 안 나옴. 문제 원인 추후 찾아보기 
n, m = map(int, input().split())
x, y, d = map(int, input().split()) # x 위치, y위치, 방향 

# 지도
gameMap = []

for _ in range(n):
    gameMap.append(list(map(int, input().split())))

directions = [(0,  -1), (1, 0), (0, 1), (-1, 0)] # 북동남서 순. 지정된 방향 값과 동일한 순서.  
directionInt = [0, 1, 2, 3] # 북, 동, 남, 서 

gameMap[x][y] = -1 # 현재 방문한 위치의 값을 -1로 변경 
# 매뉴얼에 따라 캐릭터 이동시키기
# 더이상 이동이 불가능할 때 이동을 멈추기 
while True:
    # 반시계 방향으로 회전
    d = (d + 4 - 1) % 4 # 반시계 방향으로 방향 값 회전. 항상 0~3 사이 숫자가 나올 수 있게 보정. 

    if gameMap[x + directions[d][0]][y + directions[d][1]] != 0:   # 전진할 칸이 이미 가 본 칸이거나 바다 칸
        # 현재 바라본 방향에서 뒤로 한 칸 물러나면 바다라서 더 이상 이동이 불가능할 때 이동을 멈추기
        if gameMap[x - directions[d][0]][y - directions[d][1]] == 1:
            break
        else: # 현재 바라본 방향에서 뒤로 한 칸 물러나기
            x = x - directions[d][0]
            y = y - directions[d][1]
    else:
        # 현재 바라보는 방향에서 한 칸 전진
        x += directions[d][0] 
        y += directions[d][1]

        gameMap[x][y] = -1 # 가 본 칸의 값은 -1로 변경 

# map에서 -1로 바뀐 항목 갯수 세기
print(gameMap.count(-1))
```

```python
# n, m을 공백으로 구분하여 입력 받기
n, m = map(int, input().split())

# 방문한 위치를 저장하기 위한 맵을 생성하여 0으로 초기화
d = [[0]* m for _ in range(n)]

# 현재 캐릭터의 x좌표, y좌표, 방향 입력 받기
x, y, direction = map(int, input().split())
d[x][y] = 1 # 현재 좌표 방문 자리

# 전체 맵 정보 입력 받기
array = []
for i in range(n):
    array.append(list(map(int, input().split())))

# 북, 동, 남, 서 방향 정의
dx = [-1, 0, 1, 0]
dy = [0, 1, 0, -1]

# 왼쪽으로 회전
def turn_left():
    global direction
    direction -= 1
    if direction == -1:
        direction = 3

# 시뮬레이션 시작
count = 1
turn_time = 0

while True:
    # 왼쪽으로 회전
    turn_left()
    nx = x + dx[direction]
    ny = y + dy[direction]

    # 회전한 이후 정면에 가보지 않은 칸이 존재하는 경우 이동
    if d[nx][ny] == 0 and array[nx][ny] == 0:
        d[nx][ny] = 1
        x = nx
        y = ny
        count += 1
        turn_time = 0
        continue
    # 회전한 이후 정면에 가보지 않은 칸이 없거나 바다인 경우
    else: 
        turn_time += 1
    
    # 네 방향 모두 갈 수 없는 경우
    if turn_time == 4:
        nx = x - dx[direction]
        ny = y - dy[direction]

        # 뒤로 갈 수 있다면 이동하기
        if array[nx][ny] == 0:
            x = nx
            y = ny
        # 뒤가 바다로 막혀 있는 경우
        else:
            break
        turn_time = 0

# 정답 출력
print(count)
```
