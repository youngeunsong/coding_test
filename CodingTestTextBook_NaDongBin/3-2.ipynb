# 큰 수의 법칙
다양한 수의 배열이 주어질 때 주어진 수들을 M번 더하여 가장 큰 수를 만드는 법칙. 단, 배열의 특정 인덱스에 해당하는 수가 연속해서 K번을 초과 불가능. 

예: 2, 4, 5, 4, 6 배열, M = 8, K = 3   
결과: 6 + 6 + 6 + 5 + 6 + 6 + 6 + 5 = 46  

단, 서로 다른 인덱스에 해당하는 수가 같은 경우에도 서로 다른 것으로 간주  
예: 3, 4, 3, 4, 3 배열, M = 7, K = 2
결과: 4 + 4 + 4 + 4+ 4+ 4+ 4 = 28 

# 입력 조건
첫째 줄: N, M, K (공백으로 구분)  
둘째 줄: N 개의 자연수   
K는 항상 M보다 작음. 

# 입력 예시
5 8 3  
2 4 5 4 6

# 출력 예시
46

```python
# 나의 답안 
n, m, k = map(int, input().split())
l = list(map(int, input().split()))
# print(n, m, k)
# print(l)

maxl = max(l) # 가장 큰 수 찾기 
# print(maxl)
result = 0 

if l.count(maxl) > 1: 
    result = maxl * m # 가장 큰 수가 여러 개 있다면 그냥 m만 곱하면 결과 도출 가능 
else: 
    # 두 번째로 큰 수 찾기
    l.remove(maxl)
    print(l)
    maxlRe = max(l)

    set = m // (k + 1)
    remain = m % (k + 1)

    result = set * (maxl * k + maxlRe) + remain * maxlRe

print(result)


```

```python
# 책의 모범 답안 
# n, m, k를 공백으로 구분하여 입력받기
n, m, k = map(int, input().split())

# n개의 수를 공백으로 구분하여 입력받기 
data = list(map(int, input().split()))

data.sort() # 입력받은 수 정렬
first = data[n - 1] # 가장 큰 수
second = data[n -2] # 두 번째로 큰 수 

# 가장 큰 수가 더해지는 횟수 계산
count = int(m / (k + 1)) * k
count += m % (k + 1)

result = 0
result += (count) * first # 가장 큰 수 더하기
result += (m - count) * second # 두 번째로 큰 수 더하기 

print(result) # 최종 답안 출력
```
