## 문제 설명
크기와 상관 없이 나열된 하나의 수열. 이 수를 큰 수부터 작은 수의 순서로 정렬하시오. 내림차순 정렬

## 입력 조건
* 첫째 줄: 수열에 속한 수의 개수(N). 1<= N <= 500
* 둘째 줄~N+1번째 줄: N개의 수. 수의 범위: 1 이상 100,000 이하의 자연수

## 출력 조건
* 수열을 공백을 기준으로 구분하여 출력 

## 입력 예시
```
3
15
27
12
```

## 출력 예시
27 15 12

````python
import sys
import io
input_data = """3
15
27
12"""

sys.stdin = io.StringIO(input_data)
input = sys.stdin.readline

n = int(input())
array = []
for _ in range(n):
    array.append(int(input()))

array.sort(reverse=True) # 내림차수 정렬

for i in range(n):
    print(array[i], end = ' ')
````
