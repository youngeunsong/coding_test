# 1이 될 때까지
시간 제한: 1초, 메모리 제한: 128 MB

어떠한 수 N이 1이 될 때까지 다음의 두 과정 중 하나를 반복적으로 선택하여 수행. 단, 두 번째 연산은 N이 K로 나누어 떨어질 때에만 가능 

1. N - 1
1. N / k 

N이 1이 될 때까지 필요한 1 혹은 2번 과정의 최소 횟수 구하기 

예시: N = 17, K = 4  
17 - 1 = 16  
16 /4  = 4  
4 / 4 = 1  
그러므로 총 3번 필요 

## 입력 조건
* 2 <= N <= 100,000, 2 <= K <= 100,000. N >= K 

## 입력 예시
25 5

## 출력 예시 
2

```python
# 모범 답안 
n, k = map(int, input().split())

result = 0

# N이 K 이상이라면 계속 K로 나누기
while n >= k:
    # N이 k로 나눠 떨어지지 않을 때엔 1 빼기
    while n % k != 0:
        n -= 1
        result += 1
    # K로 나누기
    n //= k
    result += 1

# 마지막으로 남은 수에 대하여 1씩 빼기
while n > 1:
    n -= 1
    result += 1

print(result)
```

```python
# 내가 작성한 코드 
n, k = map(int, input().split())

count = 0

while n != 1:
    if n % k == 0:
        n = n // k 
        count += 1
    else:
        n = n - 1
        count += 1
print(count)
```
