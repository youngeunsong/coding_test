# 인접 행렬 방식 그래프 예제

```python
INF = 9999999 # 무한의 비용 선언

# 2차원 리스트를 이용해 인접 행렬 표현: 각 노드 간 거리를 행렬로 표시. 안 이어진 노드의 거리를 INF로 처리
graph = [
    [0, 7, 5], 
    [7, 0, INF], 
    [5, INF, 0]
]

print(graph)
```

# 인접 리스트 방식 그래프 예제

```python
# 행이 3개인 2차원 리스트로 인접 리스트 표현
graph = [[] for _ in range(3)]

# 노드 0에 연결된 노드 정보 저장 (노드, 거리)
graph[0].append((1, 7))
graph[0].append((2, 5))

# 노드 1에 연결된 노드 정보 저장
graph[1].append((0, 7))

# 노드 2에 연결된 노드 정보 저장
graph[2].append((0, 5))

print(graph)
```
