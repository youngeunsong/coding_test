# 왕실의 나이트
* 시간 제한: 1초
* 메모리 제한: 128 MB

왕실 정원: 8 X 8 체스판 형상  
행 위치: 1~8  
열 위치: a~h  

나이트는 L자 형태로만 이동 가능: 수평 2칸 + 수직 1칸, 수직 2칸 + 수평 1칸 

나이트의 현재 위치에서 이동 가능한 경우의 수를 출력하기 

## 입력 조건
첫째 줄에 현재 나이트의 위치 좌표를 문자열로 제공. 열 + 행 (예: a1)

## 출력 조건
나이트가 이동할 수 있는 모든 경우의 수 

## 입력 예시
a1

## 출력 예시
2

```python
# 내가 작성한 코드 
position = input()

# 현재 위치를 열과 행의 숫자로 변형시키기
col = ord(position[0]) - ord('a')
row = int(position[1]) - 1

# 현재 위치 기준으로 모든 이동 경우의 수가 가능한 지 확인하기
dCol = [-2, 2, -1, 1]
dRow = [[1, -1], [1, -1], [2, -2], [2, -2]]

count = 0 

for c in range(len(dCol)): 
    for r in dRow[c]: 
        if col + dCol[c] < 8 and col + dCol[c] >= 0 and row + r < 8 and row + r >= 0: 
            count += 1

print(count)            
```

```python
# 모범 답안
# 현재 나이트의 위치 입력 받기
input_data = input()
row = int(input_data[1])
column = int(ord(input_data[0])) - int(ord('a')) + 1

# 나이트가 이동가능한 방향 8가지 정의
steps = [(-2, -1), (-2, 1), (2, -1), (2, 1), (-1, -2), (-1, 2), (1, -2), (1, 2)]

# 8가지 방향에 대하여 각 위치로 이동이 가능한 지 확인
result = 0
for step in steps: 
    # 이동하고자 하는 위치 확인
    next_row = row + step[0]
    next_column = column + step[1]

    # 해당 위치로 이동이 가능하다면 카운트 추가
    if next_row >= 1 and next_row <= 8 and next_column >= 1 and next_column <= 8:
        result += 1

print(result)
```
