# 순차 탐색 코드

```python
def sequential_search(n, target, array):
    # 각 원소를 하나씩  확인하며
    for i in range(n):
        # 현재의 원소가 찾는 원소와 동일하다면 
        if array[i] == target: 
            return i + 1 # 현재 위치 반환 (0부터 인덱스 시작하니 1 추가 )
print("생성할 원소 개수 입력 후 한 칸 띄고 찾을 문자열을 입력하세요")
input_data = input().split()
n = int(input_data[0]) # 원소의 갯수
target = input_data[1] # 찾을 문자열

print("앞서 적은 개수만큼 문자열을 입력하세요. 구분은 띄어쓰기 한 칸으로 합니다")
array = input().split()

# 순차 탐색 결과 입력
print(sequential_search(n, target, array))
```
