# 효율적인 화폐 구성
N가지 종류의 화폐가 있다. 이 화폐들의 개수를 최소한으로 이용해서 그 가치의 합이 M원이 되도록 하려고 한다. 사용가능한 각 화폐의 개수는 제한이 없다. 순열이 아닌 조합으로 화폐 구성의 개수를 센다. 예를 들어, 15원을 만들기 위해 2원, 3원 단위 화폐 사용 시 3원을 5개 사용하는 것이 가장 최소한의 화폐 개수이다. 

## 입력 조건
* 첫째 줄: N, M (1 <= N <= 100, 1<= M <= 10,000)
* 둘째 줄부터 N개의 줄에 각 화폐의 가치가 주어진다. 그 범위는 1 이상 10,000 이하이다. 

## 출력 조건
* M원을 만들기 위한 최소한의 화폐 개수 출력하시오
* 불가능할 때에는 -1을 출력하시오

## 입력 예시 1
```
2 15
2 
3
```

## 출력 예시 1
5

## 입력 예시 2
```
3 4
3
5
7
```

## 출력 예시 2
-1

````python
import sys
import io
input_data = """2 15
2
3"""
sys.stdin = io.StringIO(input_data)
input = sys.stdin.readline

# 입력 받기 
n, m = map(int, input().split())
k = [] # 화폐 단위 저장
for _ in range(n):
    k.append(int(input()))

# 점화식 구현
d = [10001] * (m + 1)

d[0] = 0

# 보텀업 방식으로 구현
for i in range(n): 
    # 화폐 단위 별로 확인
    for j in range(k[i], m + 1): 
        if d[j - k[i]] != 10001: # (i - k)원을 만드는 방법이 존재하는 경우
            d[j] = min(d[j], d[j - k[i]] + 1)

# 계산된 결과 출력
if d[m] == 10001: 
    print(-1)
else:
    print(d[m])
        
````
