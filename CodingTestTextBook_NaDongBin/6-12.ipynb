# 두 배열의 원소 교체 
## 제한
* 시간 제한: 2초
* 메모리 제한: 128 MB

## 문제 설명
* 두 개의 배열 A, B
    * 각 배열 크기: N
    * 배열의 원소는 모두 자연수
* 최대 K 번의 바꿔치기 연산 수행 가능: 배열 A와 배열 B에 있는 원소를 각각 한 개씩 골라 서로 바꾸기
* 최종 목표: 배열 A의 모든 원소 합이 최대가 되게 만들기 

## 입력 조건
* 첫 번째 줄: N, K 가 공백으로 구분 (1 <= N <= 100,000, 0 <= K <= N)
* 두 번째 줄: 배열 A의 원소들, 공백으로 구분된. 모든 원소는 10,000,000 미만의 자연수
* 세 번째 줄: 배열 B의 원소들, 공백으로 구분된. 모든 원소는 10,000,000 미만의 자연수

## 출력 조건
* 최대 k번의 바꿔치기 연산 후 배열 A의 모든 원소 합의 최댓값

## 입력 예시
```
5 3
1 2 5 4 3
5 5 6 6 5
```

## 출력 예시
26

````python
# 내가 작성한 답안 
# 풀이 방향: A에서 가장 작은 값 K개와 B에서 가장 큰 값 k개를 바꾸기. 단, 바꾸려는 A의 값이 B의 값보다 크다면 바꾸지 않는다. 
import sys
import io
input_data = '''5 3
1 2 5 4 3
5 5 6 6 5
'''
sys.stdin = io.StringIO(input_data)
input = sys.stdin.readline

n, k = map(int, input().split()) # 행렬 크기: n, 최대 k번의 바꿔치기 연산

a = list(map(int, input().split())) # 행렬 a
b = list(map(int, input().split())) # 행렬 b

a.sort() # A는 오름차순으로 정렬
b.sort(reverse = True) # B는 내림차순으로 정렬

# 바꿔치기 연산 수행
for i in range(k):
    if a[i] < b[i]:  # a[i]가 b[i]보다 작을 때 연산 수행
        a[i], b[i] = b[i], a[i]
    # 모범 답안과 내 답안이 거의 동일. 아래 부분만 모범 답안에 있음
    else: # A의 원소가 B의 원소보다 크거나 같을 때 반복문을 탈출 
        break 

print(sum(a)) # a의 모든 원소 합 출력 
````
