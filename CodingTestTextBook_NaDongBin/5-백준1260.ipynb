# DFS와 BFS
## 문제
그래프를 DFS로 탐색한 결과와 BFS로 탐색한 결과를 출력하는 프로그램을 작성하시오. 단, 방문할 수 있는 정점이 여러 개인 경우에는 정점 번호가 작은 것을 먼저 방문하고, 더 이상 방문할 수 있는 점이 없는 경우 종료한다. 정점 번호는 1번부터 N번까지이다.

## 입력
첫째 줄에 정점의 개수 N(1 ≤ N ≤ 1,000), 간선의 개수 M(1 ≤ M ≤ 10,000), 탐색을 시작할 정점의 번호 V가 주어진다. 다음 M개의 줄에는 간선이 연결하는 두 정점의 번호가 주어진다. 어떤 두 정점 사이에 여러 개의 간선이 있을 수 있다. 입력으로 주어지는 간선은 양방향이다.

## 출력
첫째 줄에 DFS를 수행한 결과를, 그 다음 줄에는 BFS를 수행한 결과를 출력한다. V부터 방문된 점을 순서대로 출력하면 된다.

## 예제 입력 1
```
4 5 1
1 2
1 3
1 4
2 4
3 4
```

## 예제 출력 1
```
1 2 4 3
1 2 3 4 
```

````python
# n, m, v 읽기
n, m, v = map(int, input().split())

# 그래프 값 읽기
graph = [[] for _ in range(n + 1)]

for _ in range(m):
    i, j = map(int, input().split())
    # 이 문제는 각 간선 별로 연결한 노드 싸을 제공. 그래프 형식에 맞게 변형. 
    graph[i].append(j)
    graph[j].append(i)

# print(graph)
# DFS 함수 정의 
def dfs(graph, v, visited):
    # 현재 노드를 방문 처리
    visited[v] = True
    print(v, end = ' ')
    
    # 현재 노드와 연결된 다른 노드를 재귀적으로 방문
    graph[v].sort() # 값이 작은 노드부터 방문하게 세팅 
    for i in graph[v]:
        if not visited[i]:
            dfs(graph, i, visited)

# visited 행렬 정의
visited = [False] * (n + 1) 

# 정의된 DFS 함수 호출 
dfs(graph, v, visited)
print()

# BFS 함수 정의
from collections import deque 
def bfs(graph, start, visited):
    queue = deque([start])

    # 현재 노드를 방문 처리
    visited[start] = True

    # 큐가 빌 때까지 반복
    while queue:
        # 큐에서 하나의 원소를 뽑아 출력
        v = queue.popleft()
        print(v, end = ' ')

        # 해당 원소와 연결된 나머지 원소를 모두 큐에 삽입
        graph[v].sort() # 값이 작은 노드부터 방문하게 세팅 
        for i in graph[v]:
            if not visited[i]:
                queue.append(i)
                visited[i] = True

# visited 행렬 정의
visited = [False] * (n + 1) 

# 정의된 BFS 함수 호출
bfs(graph, v, visited)
````
