# 개미 전사
메뚜기의 창고를 털어가려는 개미 전사. 개미가 만약 서로 인접한 식량 창고를 털면 메뚜기 정찰병은 알 수 있음. 

예) 식량 창고 4개: {1, 3, 1, 5}

개미는 2, 4번 창고를 선택 시 최대 식량(8개)을 약탈 가능

식량창고 N개에 대한 정보가 주어졌을 때 얻을 수 있는 식량의 최댓값을 구하시오 

## 입력 조건
* 첫째 줄: 식량 창고 N의 개수 (3 <= N <= 100)
* 둘째 줄: 공백으로 구분된 각 식량 창고가 보관 중인 식량의 개수 K (0 <= k <= 1000)

## 출력 조건 
* 개미가 얻을 수 있는 최대 식량 


```python
# 보텀업 방식의 풀이
# 점화식: a_i = max(a_(i-1), a_(i-2)+ k_i)
n = int(input())
k = list(map(int, input().split()))

d = [0] * (n + 1)
d[1] = k[0] 
d[2] = max(k[0], k[1])
for i in range(3, n + 1):
    d[i] = max(d[i - 1], d[i - 2] + k[i - 1])

# 계산된 결과 출력 
print(d[n])

```
