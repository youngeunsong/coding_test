# 알고리즘 수업 (선택 정렬 1)
## 문제
오늘도 서준이는 선택 정렬 수업 조교를 하고 있다. 아빠가 수업한 내용을 학생들이 잘 이해했는지 문제를 통해서 확인해보자.

N개의 서로 다른 양의 정수가 저장된 배열 A가 있다. 선택 정렬로 배열 A를 오름차순 정렬할 경우 K 번째 교환되는 수를 구해서 우리 서준이를 도와주자.

크기가 N인 배열에 대한 선택 정렬 의사 코드는 다음과 같다.

```
selection_sort(A[1..N]) { # A[1..N]을 오름차순 정렬한다.
    for last <- N downto 2 {
        A[1..last]중 가장 큰 수 A[i]를 찾는다
        if (last != i) then A[last] <-> A[i]  # last와 i가 서로 다르면 A[last]와 A[i]를 교환
    }
}
```

## 입력
첫째 줄에 배열 A의 크기 N(5 ≤ N ≤ 10,000), 교환 횟수 K(1 ≤ K ≤ N)가 주어진다.

다음 줄에 서로 다른 배열 A의 원소 A1, A2, ..., AN이 주어진다. (1 ≤ Ai ≤ 109)

## 출력
K 번째 교환되는 두 개의 수를 작은 수부터 한 줄에 출력한다. 교환 횟수가 K 보다 작으면 -1을 출력한다.

## 예제 입력 1
```
5 2
3 1 2 5 4
```

## 예제 출력 1
2 3

````python
# 다른 사람 코드 예시 출처: https://yuneenelife.tistory.com/44#google_vignette 
# 배열 A의 크기 N 과 교환횟수 K 를 입력받는다.
import sys
### 여기서부터 백준 제출 시 삭제
import io
input_data = '''5 2
3 1 2 5 4
'''
sys.stdin = io.StringIO(input_data)
### 여기까지 삭제 

input = sys.stdin.readline

N, K = list(map(int, input().rsplit()))
#배열 A의 원소들을 입력받는다.
lst = list(map(int, input().rsplit()))

# 선택 정렬 함수
def selectionSort(N):
    cnt = 0

    # 1. 배열의 끝부분부터 반복 
    for i in range(N - 1, 0, -1):
        last = i

        # 2. 최댓값을 찾는다
        for j in range(i - 1, -1, -1):
            if(lst[j] > lst[last]):
                last = j
        # 3. 최댓값과 swap. 만약 해당 위치가 최댓값이라면 pass
        if last != i:
            # 리스트값 스왑
            lst[i], lst[last] = lst[last], lst[i]
            cnt += 1

        # 4. 교환횟수에 도달하면 종료
        if cnt == K: 
            return (print(lst[last], lst[i]))
    return print(-1) # 교환횟수가 k보다 작으면 -1 출력 
# 선택 정렬 함수 호출
selectionSort(N)
````

````python
# 내가 작성한 코드: 오답 출력 (4 4) 
import sys

### 여기서부터 백준 제출 시 삭제
import io
input_data = '''5 2
3 1 2 5 4
'''
sys.stdin = io.StringIO(input_data)
### 여기까지 삭제 

input = sys.stdin.readline

n, k = map(int, input().split())
a = list(map(int, input().split()))

# 선택 정렬 구현 
count = 0 
oldK, newK = 0, 0
for last in range(n - 1, -1, -1): # 오른쪽 끝부터 정렬
    for i in range(last + 1):
        if last != i and a[i] > a[last]:
            a[last], a[i] = a[i], a[last]  
            last = i 
            count += 1

    if count == k:
        oldK = a[last]
        newK = a[i]
        break

print(oldK, newK)
    
````
