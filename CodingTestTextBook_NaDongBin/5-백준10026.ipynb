# 적록색약
## 문제
적록색약은 빨간색과 초록색의 차이를 거의 느끼지 못한다. 따라서, 적록색약인 사람이 보는 그림은 아닌 사람이 보는 그림과는 좀 다를 수 있다.

크기가 N×N인 그리드의 각 칸에 R(빨강), G(초록), B(파랑) 중 하나를 색칠한 그림이 있다. 그림은 몇 개의 구역으로 나뉘어져 있는데, 구역은 같은 색으로 이루어져 있다. 또, 같은 색상이 상하좌우로 인접해 있는 경우에 두 글자는 같은 구역에 속한다. (색상의 차이를 거의 느끼지 못하는 경우도 같은 색상이라 한다)

예를 들어, 그림이 아래와 같은 경우에
```
RRRBB
GGBBB
BBBRR
BBRRR
RRRRR
```
적록색약이 아닌 사람이 봤을 때 구역의 수는 총 4개이다. (빨강 2, 파랑 1, 초록 1) 하지만, 적록색약인 사람은 구역을 3개 볼 수 있다. (빨강-초록 2, 파랑 1)

그림이 입력으로 주어졌을 때, 적록색약인 사람이 봤을 때와 아닌 사람이 봤을 때 구역의 수를 구하는 프로그램을 작성하시오.

## 입력
첫째 줄에 N이 주어진다. (1 ≤ N ≤ 100)

둘째 줄부터 N개 줄에는 그림이 주어진다.

## 출력
적록색약이 아닌 사람이 봤을 때의 구역의 개수와 적록색약인 사람이 봤을 때의 구역의 수를 공백으로 구분해 출력한다.

## 예제 입력 1
```
5
RRRBB
GGBBB
BBBRR
BBRRR
RRRRR
```

## 예제 출력 1 
4 3

````python
# 내가 작성한 코드: 다른 사람의 예시와 비슷한 방식으로 접근해서 해결했기에 뿌듯하다! 
# 다른 코드 예시: https://jennnn.tistory.com/61 

import sys
### 백준 제출 시 제거
import io
input_data = """5
RRRBB
GGBBB
BBBRR
BBRRR
RRRRR
"""
sys.stdin = io.StringIO(input_data)
### 여기까지 작성 후 제거 
from collections import deque 
# bfs 함수 정의
dx = [-1, 1, 0, 0]
dy = [0, 0, -1, 1]

def bfs(x, y, c): # x, y 는 정수, c는 char (R, G, B)
    queue = deque()
    queue.append((x, y))
    visited[x][y] = True

    while queue: 
        x, y = queue.popleft()

        for i in range(len(dx)):
            nx = x + dx[i]
            ny = y + dy[i]

            # 그리드 범위 안에 없으면 패스 
            if nx >= n or nx < 0 or ny >= n or ny < 0:
                continue 

            # 아직 방문 안 했고, 특정 색상이 맞다면 방문 처리 및 큐에 추가
            if graph[nx][ny] == c and not visited[nx][ny]:
                visited[nx][ny] = True
                queue.append((nx, ny))

input = sys.stdin.readline

# 그리드 크기 읽기
n = int(input())

# 그래프 읽기 
graph = []

for _ in range(n):
    graph.append(list(input()))

# 적록색약이 아닌 사람: R, G, B 별로 구역의 수 카운트하기 
visited = [[False] * n for _ in range(n)]
count = 0 
for i in range(n):
    for j in range(n):
        for c in ['R', 'G', 'B']: 
            if graph[i][j] == c and not visited[i][j]:
                bfs(i, j, c)
                count += 1
print(count, end = " ")

# 적록색약인 사람: R+G, B 조합으로 카운트 
visited = [[False] * n for _ in range(n)]
count = 0 
# graph에서 G를 R로 치환하기
for i in range(n):
    for j in range(n):
        if graph[i][j] == 'G':
            graph[i][j] = 'R'

for i in range(n):
    for j in range(n): 
        if not visited[i][j]: 
            if graph[i][j] == 'R' or graph[i][j] == 'G':
                c = graph[i][j]
                bfs(i, j, c)
                count += 1
            elif graph[i][j] == 'B':
                c = graph[i][j]
                bfs(i, j, c)
                count += 1
print(count)

````
