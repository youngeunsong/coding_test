# 그리디 알고리즘
# 문제 설명
당신은 계산 담당 음식점 직원. 카운터에 거스름돈으로 사용 가능한 500, 100, 50, 10원 짜리 동전이 무한히 존재한다고 가정. 

손님에게 거슬러줘야 할 돈이 N원일 때 거슬러줘야 할 동전의 최소 갯수를 구하시오. 단, 거슬러줘야할 돈 N은 언제나 10의 배수. 

# 입력 예시
N = 1260

# 출력 예시
(500원, 100원, 50원, 10원 갯수 순으로 나열)  
2 2 1 1 

```python
n = int(input())
COINS = [500, 100, 50, 10] # 큰 화폐부터 확인 
count = 0 
for c in COINS:
    count += n // c # 해당 화폐로 거슬러 줄 수 있는 동전 갯수 세기 
    n = n % c 

print(count)    
```
