# 단지번호붙이기
## 문제
<그림 1>과 같이 정사각형 모양의 지도가 있다. 1은 집이 있는 곳을, 0은 집이 없는 곳을 나타낸다. 철수는 이 지도를 가지고 연결된 집의 모임인 단지를 정의하고, 단지에 번호를 붙이려 한다. 여기서 연결되었다는 것은 어떤 집이 좌우, 혹은 아래위로 다른 집이 있는 경우를 말한다. 대각선상에 집이 있는 경우는 연결된 것이 아니다. <그림 2>는 <그림 1>을 단지별로 번호를 붙인 것이다. 지도를 입력하여 단지수를 출력하고, 각 단지에 속하는 집의 수를 오름차순으로 정렬하여 출력하는 프로그램을 작성하시오.




## 입력
첫 번째 줄에는 지도의 크기 N(정사각형이므로 가로와 세로의 크기는 같으며 5≤N≤25)이 입력되고, 그 다음 N줄에는 각각 N개의 자료(0혹은 1)가 입력된다.

## 출력
첫 번째 줄에는 총 단지수를 출력하시오. 그리고 각 단지내 집의 수를 오름차순으로 정렬하여 한 줄에 하나씩 출력하시오.

## 예제 입력 1
```
7
0110100
0110101
1110101
0000111
0100000
0111110
0111000
```
## 예제 출력 1
```
3
7
8
9
```

````python
# 출처: https://hongcoding.tistory.com/71
import sys
sys.setrecursionlimit(10000) # 재귀함수 상한선 해제
### 여기서부턴 백준 제출 시 제거
import io
input_data = """7
0110100
0110101
1110101
0000111
0100000
0111110
0111000
"""
sys.stdin = io.StringIO(input_data)
### 여기까지 제거
input = sys.stdin.readline

# 지도의 크기 읽기
n = int(input())

# 지도 읽기
graph = []
for _ in range(n):
    graph.append(list(map(int, input().strip())))

# 상하좌우 파악용 상수 
dx = [-1, 1, 0, 0]
dy = [0, 0, -1, 1]

# dfs 함수 정의
def dfs(x, y):
    # 지도 범위 밖이면 종료 
    if x >= n or x < 0 or y >= n or y < 0:
        return False

    # 노드 방문 처리
    if graph[x][y] == 1:
        global count 
        count += 1
        graph[x][y] = 0 

        # 인접 노드 방문 가능여부 확인
        for i in range(4):
            nx = x + dx[i]
            ny = y + dy[i]
            dfs(nx, ny)
        return True
    return False 

count = 0
result = 0 # 단지수 
num = []

for i in range(n):
    for j in range(n):
        if dfs(i, j) is True:
            num.append(count)
            result += 1
            count = 0
num.sort() # 단지 별 아파트 수 오름차순으로 정렬 
print(result) # 단지 수 출력
for i in range(len(num)):
    print(num[i]) # 단지 별 아파트 수 출력
````
