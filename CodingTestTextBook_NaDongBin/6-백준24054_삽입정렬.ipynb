# 알고리즘 수업 - 삽입 정렬 4
## 문제
오늘도 서준이는 삽입 정렬 수업 조교를 하고 있다. 아빠가 수업한 내용을 학생들이 잘 이해했는지 문제를 통해서 확인해보자.

N개의 서로 다른 양의 정수가 저장된 배열 A가 있다. 삽입 정렬로 배열 A를 오름차순 정렬할 경우 K 번째 저장되는 수를 구하자.

N이 매우 커서 시간 초과를 고민하고 있는 우리 서준이를 도와주자.

크기가 N인 배열에 대한 삽입 정렬 의사 코드는 다음과 같다.

```
insertion_sort(A[1..N]) { # A[1..N]을 오름차순 정렬한다.
    for i <- 2 to N {
        loc = i - 1;
        newItem = A[i];

        # 이 지점에서 A[1..i-1]은 이미 정렬되어 있는 상태
        while (1 ≤ loc and newItem < A[loc]) {
            A[loc + 1] <- A[loc];
            loc--;
        }
        if (loc + 1 != i) then A[loc + 1] = newItem;
    }
}
```

## 입력
첫째 줄에 배열 A의 크기 N(5 ≤ N ≤ 500,000), 저장 횟수 K(1 ≤ K ≤ 1012)가 주어진다.

다음 줄에 서로 다른 배열 A의 원소 A1, A2, ..., AN이 주어진다. (1 ≤ Ai ≤ 109)

## 출력
K 번째 저장 되는 수를 출력한다. 저장 횟수가 K 보다 작으면 -1을 출력한다.

## 예제 입력 1 
```
5 7
4 5 1 3 2
```

## 예제 출력 1 
5

````python
# 시간 초과 오류 발생. pypy3 적용해도 발생 -> 집에 가서 다시 제출 해보기 

import sys

### 백준 제출 시 제거
import io
input_data = '''5 7
4 5 1 3 2'''
sys.stdin = io.StringIO(input_data)
###
input = sys.stdin.readline

# 입력 받기
n, k = map(int, input().split())
a = list(map(int, input().split()))

# 의사 코드 
# insertion_sort(A[1..N]) { # A[1..N]을 오름차순 정렬한다.
#     for i <- 2 to N {
#         loc = i - 1;
#         newItem = A[i];

#         # 이 지점에서 A[1..i-1]은 이미 정렬되어 있는 상태
#         while (1 ≤ loc and newItem < A[loc]) {
#             A[loc + 1] <- A[loc];
#             loc--;
#         }
#         if (loc + 1 != i) then A[loc + 1] = newItem;
#     }
# }

# 삽입 정렬 함수 구현
def insertion_sort(a):
    count = 0

    for i in range(1, n):
        loc = i - 1
        newItem = a[i]

        while 0 <= loc and newItem < a[loc]:
            a[loc + 1] = a[loc]
            loc -= 1
            count += 1
            # print(count, 0, a)

            if count == k:
                return print(a[loc + 1])
        
        if loc + 1 != i:
            a[loc + 1] = newItem
            count += 1
            # print(count, 1, a)

            if count == k:
                return print(a[loc + 1])
    print(-1)

insertion_sort(a)

````
