```python
# 9506: 약수들의 합
while True:
    n = int(input())
    
    # 마지막 입력값 받으면 루프 끝냄. 
    if n == -1:
        break

    # 약수 목록 구하기
    divisors = []

    for i in range(1, n):
        if n % i == 0:
            divisors.append(i)
    
    # 완전수인지 판별
    if n == sum(divisors):   
        fin_sent = str(n) + " = "
        for i in divisors:
            if i == 1:
                fin_sent += str(i)
            else:
                fin_sent += " + " + str(i)
        print(fin_sent)
                
    else:
        print(n, "is NOT perfect.")
```

```python
#2309: 일곱난쟁이
arr = []

# 9명의 난쟁이 키 받기
for _ in range(9):
    arr.append(int(input()))


# 조합. itertools없이 구현해보기
def comb(arr, n):
    result = []

    if n > len(arr):
        return result
    
    if n == 1:
        for i in arr:
            result.append([i])
    elif n > 1:
        for i in range(len(arr) - n + 1):
            for j in comb(arr[i+1:], n - 1):
                result.append([arr[i]] + j)
                # print([arr[i]], j)
    return result

combinations = comb(arr, 7)

for combi in combinations:
    if sum(combi) == 100: 
        combi.sort()

        for i in range(len(combi)):
            print(combi[i])

        break

# arr = [1, 2, 3]
# print(comb(arr, 2))

# arr = [20, 7, 23, 19, 10, 15, 25, 8, 13]
# print(comb(arr, 7))
```

```python
# 조합을 재귀 함수로 다시 한 번 구현
# 참조: https://cotak.tistory.com/70
def get_combinations(arr, n):
    result = []

    if n == 0:
        return [[]]
    
    for i in range(len(arr)):
        element = arr[i]
        rest_arr = arr[i+1:]

        for j in get_combinations(rest_arr, n-1):
            result.append([element] + j)
            # print([element])
            # print(j)
    return result
```

```python
get_combinations([1, 2, 3, 4], 3)
```

```python
# 순열을 재귀함수로 구현
def get_permutations(arr, n):
    result = []

    if n == 0:
        return [[]]
    
    for i in range(len(arr)):
        element = arr[i]
        rest_arr = arr[:i] + arr[i+1:]

        for j in get_permutations(rest_arr, n - 1):
            result.append([element] + j)

    return result
```

```python
get_permutations([0, 1, 2, 3], 2)
```

```python
# 2609: 최대공약수와 최소공배수
# 유클리드 호제법으로 최대공약수와 최소공배수 구해보기
n, m = map(int, input().split())

def get_gcd(x, y):
    if x % y == 0:
        return y
    
    return(get_gcd(y, x % y))

def get_lcm(x, y):
    gcd = get_gcd(x, y)
    return gcd * (x // gcd) * (y // gcd) 

print(get_gcd(n, m))
print(get_lcm(n, m))
```

```python
#1978: 소수 찾기
def is_prime(n):
    if n == 1:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

n = int(input())
arr = list(map(int, input().split()))
count = 0
for i in range(n):
    if is_prime(arr[i]):
        count += 1
print(count)

```

```python
# 2581: 소수
def is_prime(n):
    if n == 1:
        return False
    
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

m = int(input())
n = int(input())
sum_primes = 0
min_prime = n

for i in range(m, n + 1):
    if is_prime(i):
        sum_primes += i
        
        if i < min_prime:
            min_prime = i

if sum_primes > 0:
    print(sum_primes)
    print(min_prime)
else:
    print(-1)
```

## 준비운동 part2: 약점 체크
재귀 탐색의 기본: 연산자 끼워넣기

[https://data-flower.tistory.com/72]

```python
# 14888: 연산자 끼워넣기
# 참조: https://data-flower.tistory.com/72
n = int(input()) # 수의 개수
data = list(map(int, input().split())) # 수열
add, sub, mul, div = map(int, input().split()) # 연산자 개수 계산

# 최댓값과 최솟값 초기화
max_val = -1e9
min_val = 1e9

# dfs 매서드 정의
def dfs(i, arr):
    global add, sub, mul, div, max_val, min_val

    # 주어진 수열을 다 받았을 경우 최댓값과 최솟값 계산
    if i == n:
        max_val = max(max_val, arr)
        min_val = min(min_val, arr)
    else:
        # 더하기
        if add > 0:
            add -= 1
            dfs(i + 1, arr + data[i])  
            # print(add, i, arr)
            add += 1 # 이걸 없애면 작동 안함. 
        # 빼기
        if sub > 0:
            sub -= 1
            dfs(i + 1, arr - data[i])
            # print(sub, i, arr)
            sub += 1    
        # 곱하기
        if mul > 0:
            mul -= 1
            dfs(i + 1, arr * data[i])
            # print(mul, i, arr)
            mul += 1
        # 나누기
        if div > 0:
            div -= 1
            # 음수를 양수로 나눌 때는 C++14의 기준을 따른다. 즉, 양수로 바꾼 뒤 몫을 취하고, 그 몫을 음수로 바꾼 것과 같다
            if arr >= 0:
                dfs(i + 1, arr // data[i])
            else:
                dfs(i + 1, -(abs(arr) // data[i]))

            # print(div, i, arr)
            div += 1

# DFS 메서드 호출
dfs(1, data[0])

# 최댓값과 최솟값 호출
print(max_val)
print(min_val)
```

스택의 응용 - 2504: 괄호의 값

예제로 룰 이해해보기
(()[[]])([])

1. (**()[[]]**)([]) : 2 + 3 * 3 = 11
1. **(()[[]])**([]) : 2 * 11 = 22
1. (()[[]])**([])** : 2 * 3 = 6
1. 그러므로 최종 값은 22 + 6 = 28
괄호 안에 있는 관계는 곱셉, 그렇지 않으면 덧셈으로 연결되어 있음. 

[https://jie0025.tistory.com/175]

```python
inputs = input()
stack = []
tmp = 1
answer = 0

for i in range(len(inputs)):
    bracket = inputs[i]
    if bracket == '(':
        tmp *= 2
        stack.append(bracket)
    elif bracket == '[':
        tmp *= 3
        stack.append(bracket)
    elif bracket == ')':
        if not stack or stack[-1] == '[':
            answer = 0
            break
        if inputs[i-1] == '(':
            answer += tmp
        tmp //= 2
        stack.pop()
    else:
        if not stack or stack[-1] == '(':
            answer = 0
            break   
        if inputs[i-1] == '[':
            answer += tmp
        tmp //= 3
        stack.pop()

if stack:
    print(0)
else:
    print(answer)

```

시뮬레이션 기본 - 빗물 14719번
시뮬레이션(Simulation): 일련의 명령에 따라서 개체를 차례대로 이동시키는 것

풀이 방법: 
특정 위치에 물이 고이는지 안 고이는지 - 자신보다 높은 블럭이 왼쪽과 오른쪽에 모두 있는지 판별. 
물이 고이는 양 = 왼쪽과 오른쪽 블록 중 더 낮은 블록과 현 위치의 차. 
[https://seongonion.tistory.com/115]

```python
h, w = map(int, input().split())
heights = list(map(int, input().split()))
answer = 0

for i in range(1, w-1):
    left, right = max(heights[:i]), max(heights[i+1:])

    if heights[i] < min(left, right):
        answer +=  min(left, right) - heights[i]

print(answer)
```

완전탐색의 유연한 생각: 가르침 - 1062번  
알고리즘 분류
- 브루트포스 알고리즘
- 비트마스킹
- 백트래킹  

문제 이해:
k개의 글자로만 이루어진 단어. 남극 언어는 단어가 N개. 시작 = anta, 끝 = tica  

풀이 방법:  
anta, tica가 무조건 들어가는 단어들 -> a, c, i, n, t라는 5개의 알파벳은 무조건 읽을 줄 알아야함. 안 그러면 단어를 1개도 못 읽음. 즉, k가 5개 미만이면 애초에 아무 단어도 못 읽는 상황. 

만약 k가 5이상일 때, 각 단어 별로 anta와 tica 사이에 사용된 글자 중 최대한 많은 단어를 커버 가능한 글자를 선택할 수 있는지, 그리고 그 떄 커버 가능한 단어 수를 구하면 되는 문제. -> k-5개의 알파벳에 대한 DFS

풀이 참조: <https://kyun2da.github.io/2020/09/26/teaching/> -> 근데 시간 초과 뜸

```python
# 시간초과가 뜨는 방식

n, k = map(int, input().split())

# k가 5 미만이면 그 어떤 단어도 배울 수 없음. 
if k < 5:
    print(0)
    exit()

# k가 26이면 모든 단어를 배울 수 있음
elif k == 26:
    print(n)
    exit()

answer = 0
words = [input() for _ in range(n)]
# print(words)
learn = [0] * 26

# 적어도 단어 하나는 배우기 위해 a, c, i, n, t는 무조건 배워야 함. 
for c in ('a', 'c', 'i', 'n', 't'):
    learn[ord(c) - ord('a')] = 1

def dfs(idx, cnt):
    global answer

    if cnt == k - 5:
        readcnt = 0
        for word in words:
            check = True
            for w in word:
                if not learn[ord(w) - ord('a')]:
                    check = False
                    break
            if check:
                readcnt += 1
        answer = max(answer, readcnt)
        return
    
    for i in range(idx, 26):
        if not learn[i]:
            learn[i] = True
            dfs(i, cnt + 1)
            learn[i] = False

dfs(0, 0)
print(answer)


```

```python
# 스스로 연습
n, k = map(int, input().split())

# k 가 5 미만이면 단어를 절대 배울 수 없음
if k < 5:
    print(0)
    exit()

# k = 26이면 모든 단어를 배울 수 있음
elif k == 26:
    print(n)
    exit()

# 5 <= k < 26일 때의 경우
answer = 0
words = []
for _ in range(n):
    words.append(input())
learn = [0] * 26 # 각 알파벳 별로 배웠으면 1(True), 아니면 0(False)

# 무조건 단어에 쓰이는 'a', 'c', 'i', 'n', 't'는 배운다고 가정한다
for c in ('a', 'c', 'i', 'n', 't'):
    learn[ord(c) - ord('a')] = 1

# dfs로 배울 수 있는 단어가 최대가 되는 경우 찾기
def dfs(idx, cnt): # idx : learn의 인덱스
    global answer

    if cnt == k - 5: # 탐색 종료
        readcnt = 0
        for word in words:
            check = True
            for w in word:
                if not learn[ord(w) - ord('a')]:
                    check = False
                    break
            if check:
                readcnt += 1 # 이 단어는 읽을 수 있다. 

        answer = max(answer, readcnt)
        return 
    
    # 검색
    for i in range(idx, 26): #idx부터 z까지 알파벳 중 탐색
        if not learn[i]:
            learn[i] = True
            dfs(i, cnt + 1) # 재귀로 탐색. 즉 
            learn[i] = False

dfs(0, 0)
print(answer)


    
```
